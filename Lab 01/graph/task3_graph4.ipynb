```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
t = pd.read_excel("RunningTimeRecord.xlsx", sheet_name="Testing 2", skiprows=3)
```

```python
t2 = t.iloc[0:6, [0,2,4]].copy()

t2.rename(
    columns={
        "Num Threads": "num_t",
        "Computational Time (sec).1": "posix_comp_time",
        "Computational Time (sec).2": "openmp_comp_time",},
    inplace=True)

t2
```

```python
t3 = t.iloc[11:18, [0,2,4]].copy()

t3.rename(
    columns={
        "Num Threads": "num_t",
        "Computational Time (sec).1": "posix_comp_time",
        "Computational Time (sec).2": "openmp_comp_time",},
    inplace=True)

t3.reset_index(drop=True, inplace=True)

t3
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(16, 6), dpi=300)


# =============================================================
# 8-core graph
# =============================================================
ax = axes[0]

# Use index positions for equal spacing
x2 = range(len(t2.num_t))

line1, = ax.plot(
    x2, t2.posix_comp_time,
    marker='o',
    markersize=6,
    linewidth=2.2,
    color='#0275D8',
    label='POSIX Threads'
)

line2, = ax.plot(
    x2, t2.openmp_comp_time,
    marker='o',
    markersize=6,
    linewidth=2.2,
    color='#0FDE00',
    label='Open MP'
)

ax.set_title(
    "Testing on 8-core Machine",
    fontsize=12,
    loc='left'
)

ax.set_xlabel(
    "Number of Threads",
    fontsize=12,
    fontweight='semibold',
    labelpad=10
)

ax.set_ylabel(
    "Computational Time (seconds)",
    fontsize=12,
    fontweight='semibold',
    labelpad=10
)

# X-axis labels
ax.set_xticks(range(len(t2.num_t)))
ax.set_xticklabels(t2.num_t, fontsize=10)

ax.tick_params(axis='y', labelsize=10)

ax.grid(True, linestyle='--', alpha=0.6)
ax.set_ylim(-0.5, 3)


# -------------------------------------------------------------
# Annotations - POSIX Threads
# -------------------------------------------------------------

# First point
ax.annotate(
    f"{t2.posix_comp_time.iloc[0]:.2f}s",
    (0, t2.posix_comp_time.iloc[0]),
    textcoords='offset points',
    xytext=(0, -16),
    ha='center',
    fontsize=8.5,
    color='#014C8C'
)

# Last point
ax.annotate(
    f"{t2.posix_comp_time.iloc[-1]:.2f}s",
    (len(t2.num_t) - 1, t2.posix_comp_time.iloc[-1]),
    textcoords='offset points',
    xytext=(0, 7),
    ha='center',
    fontsize=8.5,
    color='#014C8C'
)


# -------------------------------------------------------------
# Annotations - Open MP
# -------------------------------------------------------------

# First point
ax.annotate(
    f"{t2.openmp_comp_time.iloc[0]:.2f}s",
    (0, t2.openmp_comp_time.iloc[0]),
    textcoords='offset points',
    xytext=(0, 7),
    ha='center',
    fontsize=8.5,
    color='#058F03'
)

# Last point
ax.annotate(
    f"{t2.openmp_comp_time.iloc[-1]:.2f}s",
    (len(t2.num_t) - 1, t2.openmp_comp_time.iloc[-1]),
    textcoords='offset points',
    xytext=(0, -14),
    ha='center',
    fontsize=8.5,
    color='#058F03'
)


# =============================================================
# 32-core graph
# =============================================================
ax = axes[1]

# Use index positions for equal spacing
x3 = range(len(t3.num_t))

ax.plot(
    x3, t3.posix_comp_time,
    marker='o',
    markersize=6,
    linewidth=2.2,
    color='#0275D8',
    label='POSIX Threads'
)

ax.plot(
    x3, t3.openmp_comp_time,
    marker='o',
    markersize=6,
    linewidth=2.2,
    color='#0FDE00',
    label='Open MP'
)

ax.set_title(
    "Testing on 32-core Machine",
    fontsize=12,
    loc='left'
)

ax.set_xlabel(
    "Number of Threads",
    fontsize=12,
    fontweight='semibold',
    labelpad=10
)

# X-axis labels
ax.set_xticks(range(len(t3.num_t)))
ax.set_xticklabels(t3.num_t, fontsize=10)

ax.tick_params(axis='y', labelsize=10)

ax.grid(True, linestyle='--', alpha=0.6)
ax.set_ylim(-0.5, 3)


# -------------------------------------------------------------
# Annotations - POSIX Threads
# -------------------------------------------------------------

# First point
ax.annotate(
    f"{t3.posix_comp_time.iloc[0]:.2f}s",
    (0, t3.posix_comp_time.iloc[0]),
    textcoords='offset points',
    xytext=(0, -16),
    ha='center',
    fontsize=8.5,
    color='#014C8C'
)

# Last point
ax.annotate(
    f"{t3.posix_comp_time.iloc[-1]:.2f}s",
    (len(t3.num_t) - 1, t3.posix_comp_time.iloc[-1]),
    textcoords='offset points',
    xytext=(0, 7),
    ha='center',
    fontsize=8.5,
    color='#014C8C'
)


# -------------------------------------------------------------
# Annotations - Open MP
# -------------------------------------------------------------

# First point
ax.annotate(
    f"{t3.openmp_comp_time.iloc[0]:.2f}s",
    (0, t3.openmp_comp_time.iloc[0]),
    textcoords='offset points',
    xytext=(0, 7),
    ha='center',
    fontsize=8.5,
    color='#058F03'
)

# Last point
ax.annotate(
    f"{t3.openmp_comp_time.iloc[-1]:.2f}s",
    (len(t3.num_t) - 1, t3.openmp_comp_time.iloc[-1]),
    textcoords='offset points',
    xytext=(0, -14),
    ha='center',
    fontsize=8.5,
    color='#058F03'
)


# =============================================================
# Overall title
# =============================================================
fig.suptitle(
    "POSIX Threads vs. Open MP with Number of Threads",
    fontsize=16,
    fontweight='bold',
    x=0.055,
    y=0.92,
    ha='left'
)


# =============================================================
# Overall legend
# =============================================================
fig.legend(
    handles=[line1, line2],
    labels=['POSIX Threads', 'Open MP'],
    loc='upper right',
    bbox_to_anchor=(0.994, 0.94),
    ncol=2,
    fontsize=10.5,
    frameon=True,
    framealpha=0.9,
    shadow=True
)


# =============================================================
# Layout & save
# =============================================================
plt.tight_layout(rect=[0, 0, 1, 0.94])

plt.savefig("task3_graph4.png", dpi=300, bbox_inches='tight')

plt.show()
```
